```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('students.csv')

# Histogram
sns.histplot(df['Marks'], kde=True)
plt.title("Marks distribution")
plt.show()

# Box plot
sns.boxplot(x='Branch', y='Marks', data=df)
plt.title("Marks by branch")
plt.show()

# Scatter plot
sns.scatterplot(x='StudyHours', y='Marks', hue='Result', data=df)
plt.title("Study hours V/S Marks")
plt.show()

# Pair plot
sns.pairplot(df[['StudyHours', 'Attendance', 'PrevMarks', 'Marks', 'Result']], hue='Result')
plt.show()

# Calculate the correlation matrix for specific columns
corr = df[["StudyHours", "Attendance", "PrevMarks", "Marks"]].corr()

# Generate the correlation heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")

# Add title and display the plot
plt.title("correlation Heatmap")
plt.show()
```
