```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# 1. Construct representative dummy dataset
df = pd.DataFrame(
    {
        "Income": [50000, np.nan, 48000, 120000, 52000],  # Numerical (Skewed)
        "Age": [25, 30, np.nan, 35, 28],  # Numerical (Normal)
        "Satisfaction": [
            "Medium",
            "Low",
            "High",
            np.nan,
            "Medium",
        ],  # Categorical (Ordinal)
        "Device": ["Mobile", "Desktop", "Mobile", np.nan, "Tablet"],  # Nominal
    }
)

# --- SECTION A: HANDLING MISSING VALUES ---

# Mean Imputation for Normally Distributed Features (Age)
mean_imp = SimpleImputer(strategy="mean")
df["Age"] = mean_imp.fit_transform(df[["Age"]])

# Median Imputation for Outlier-Prone / Skewed Features (Income)
median_imp = SimpleImputer(strategy="median")
df["Income"] = median_imp.fit_transform(df[["Income"]])

# Mode Imputation for Categorical Features prior to Encoding
mode_imp = SimpleImputer(strategy="most_frequent")
df[["Satisfaction", "Device"]] = mode_imp.fit_transform(
    df[["Satisfaction", "Device"]]
)

# Forward Fill Alternative (Commonly used for pure time-series tracks)
# df['Sensor_Data'] = df['Sensor_Data'].ffill()

# --- SECTION B: ENCODING CATEGORICAL VARIABLES ---

# Ordinal Encoding (Preserving hierarchical rank)
rank_order = [["Low", "Medium", "High"]]
ord_enc = OrdinalEncoder(categories=rank_order)
df["Satisfaction_Encoded"] = ord_enc.fit_transform(df[["Satisfaction"]])

# One-Hot Encoding (For nominal, unordered features)
ohe = OneHotEncoder(sparse_output=False, drop="first")
device_ohe = ohe.fit_transform(df[["Device"]])
device_df = pd.DataFrame(
    device_ohe, columns=ohe.get_feature_names_out(["Device"])
)

# Combine results into final clean array
final_df = pd.concat(
    [df.drop(columns=["Satisfaction", "Device"]), device_df], axis=1
)
print(final_df)
```
