```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [20, 25, np.nan, 30, 35],
    'Salary': [20000, np.nan, 30000, 40000, 50000],
    'Experience': [1, np.nan, 3, np.nan, 5],
    'City': ['Delhi', 'Mumbai', np.nan, 'Delhi', 'Mumbai']
})

# Mean imputation
df['Age_Mean'] = df['Age'].fillna(df['Age'].mean())

# Median imputation
df['Salary_Median'] = df['Salary'].fillna(df['Salary'].median())

# Forward fill
df['Experience_ffill'] = df['Experience'].ffill()

# Backward fill
df['Experience_bfill'] = df['Experience'].bfill()

# Mode imputation
df['City_Mode'] = df['City'].fillna(df['City'].mode()[0])

print(df)
```
