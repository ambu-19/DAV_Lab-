```python
import numpy as np
import pandas as pd

# Read CSV file
df = pd.read_csv('World Economic Indicators - Sheet1.csv')

# Calculate Q1 and Q3
Q1 = df['Population density'].quantile(0.25)
Q3 = df['Population density'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Find outliers
outliers_iqr = df[
    (df['Population density'] < lower) |
    (df['Population density'] > upper)
]

# Display results
print("Lower bound:", lower)
print("Upper bound:", upper)
print(f"IQR bounds: [{lower:.2f}, {upper:.2f}]")
print(outliers_iqr)
print("Outliers found (IQR):", len(outliers_iqr))
```

```python
# Z-score method means it is a statistical measurement that shows how many stds a datapoint lies above or below the mean of distribution

df['Z_Score']=(df['Population density']-df['Population density'].mean()) / df['Population density'].std()
outliers_z = df[df['Z_Score'].abs()>3]
print("Z_Score>3 :",len(outliers_z))
```

```python

df['Population density_capping']=df['Population density'].clip(lower=lower,upper=upper)
print("Before:",df['Population density'].max())
print("After:",df['Population density_capping'].max())
```
