```python
import numpy as np
import pandas as pd
df=pd.read_csv('students.csv')
Q1=df['StudyHours'].quantile(0.25)
Q3=df['StudyHours'].quantile(0.75)
IQR = Q3-Q1
lower = Q1-1.5*IQR
upper= Q3+1.5*IQR
outliers_iqr=df[(df['StudyHours'] < lower) | (df['StudyHours'] > upper)]
print(lower)
print(upper)
print(f'IQR bounds:[{lower:.2f}, {upper:.2f}]')
print(outliers_iqr)
print("Outliers found (IRQ):",len(outliers_iqr))
```

```python
# Z-score method means it is a statistical measurement that shows how many stds a datapoint lies above or below the mean of distribution

df['Z_Score']=(df['StudyHours']-df['StudyHours'].mean()) / df['StudyHours'].std()
outliers_z = df[df['Z_Score'].abs()>3]
print("Z_Score>3 :",len(outliers_z))
```

```python
df['StudyHours_capping']=df['StudyHours'].clip(lower=lower,upper=upper)
print("Before:",df['StudyHours'].max())
print("After:",df['StudyHours_capping'].max())
```
